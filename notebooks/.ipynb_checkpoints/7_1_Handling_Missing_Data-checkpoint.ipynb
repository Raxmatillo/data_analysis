## 7.1 Handling Missing Data

```python
import numpy as np
import pandas as pd
```

```python
float_data = pd.Series([1.2, -3.5, np.nan, 0])
float_data
```

```python
float_data.isna()
```

```python
string_data = pd.Series(["aardvark", np.nan, None, "avocado"])
string_data
```

```python
string_data.isna()
```

```python
string_data.isnull()
```

```python
float_data = pd.Series([1, 2, None], dtype='float64')
float_data
```

```python
float_data.isna()
```

### Filtering Out Missing Data

```python
data = pd.Series([1, np.nan, 3.5, np.nan, 7])
data.dropna()
```

```python
data[data.notna()]
```

```python
data = pd.DataFrame([[1., 6.5, 3.], [1., np.nan, np.nan],
                     [np.nan, np.nan, np.nan], [np.nan, 6.5, 3.]])

data
```

```python
data.dropna()
```

```python
data.dropna(how="all")
```

```python
data[4] = np.nan
data
```

```python
data.dropna(axis="columns", how="all")
```

```python
df = pd.DataFrame(np.random.standard_normal((7, 3)))
```

```python
df.iloc[:4, 1] = np.nan
df.iloc[:2, 2] = np.nan
df
```

```python
df.dropna()
```

```python
df.dropna(thresh=2)
```

### Filling In Missing Data

```python
df.fillna(0)
```

```python
df.fillna({1:0.5, 2:0})
```

```python
df = pd.DataFrame(np.random.standard_normal((6, 3)))
```

```python
df.iloc[2:, 1] = np.nan
df.iloc[4:, 2] = np.nan
```

```python
df
```

```python
df.ffill()
```

```python
df.ffill(limit=2)
```

```python
data = pd.Series([1., np.nan, 3.5, np.nan, 7])
```

```python
data.fillna(data.mean())
```
